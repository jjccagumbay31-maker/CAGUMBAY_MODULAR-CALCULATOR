```python
# Simple Python Calculator
print("=== Simple Python Calculator ===")

# Get two numbers from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Display menu options
print("\nChoose an operation:")
print("1. Add (+)")
print("2. Subtract (-)")
print("3. Multiply (*)")
print("4. Divide (/)")

# Get the user's choice
choice = input("Enter your choice (1/2/3/4): ")

print("\n-------------------------")

# Perform calculation based on choice
if choice == '1':
    result = num1 + num2
    print("Result: ", num1, " + ", num2, " = ", result)
elif choice == '2':
    result = num1 - num2
    print("Result: ", num1, " - ", num2, " = ", result)
elif choice == '3':
    result = num1 * num2
    print("Result: ", num1, " * ", num2, " = ", result)
elif choice == '4':
    # Prevent division by zero
    if num2 == 0:
        print("Error: Cannot divide by zero.")
    else:
        result = num1 / num2
        print("Result: ", num1, " / ", num2, " = ", result)
else:
    print("Invalid choice. Please run the program again and select 1-4.")

print("-------------------------")
#Reflection#
#1. The functions i used was user definded functions and defined functions
#2  I used keyword parameters and positional parameters
#3. No arguments were passed
#4. used input and float such as num 1,num2,num3 being stored and reused
#5. For Reusability, Readability, and Debugging
```
